## Exercise 1: Prime Numbers

**Problem:**  
Write a Python program that checks whether a given number is prime or not.

A prime number is a natural number greater than 1 that has no positive divisors other than 1 and itself.


````python
# Exercise 1: Prime Number Check

# Taking input from the user
num = int(input("Enter a number: "))

# Prime numbers are greater than 1
if num <= 1:
    print(num, "is not a prime number")
else:
    is_prime = True

    # Checking divisibility from 2 to square root of num
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            is_prime = False
            break

    if is_prime:
        print(num, "is a prime number")
    else:
        print(num, "is not a prime number")
````

### Explanation

- A number less than or equal to 1 is not prime.
- We check whether the number is divisible by any number from 2 to square root of the number.
- If it is divisible, it is not prime.
- Otherwise, it is prime.


## Exercise 2: Product of Random Numbers

**Problem:**  
Develop a Python program that generates two random numbers and asks the user to enter the product of these numbers.  
The program should check if the user's answer is correct and display an appropriate message.


````python
# Exercise 2: Product of Random Numbers

import random

# Generate two random numbers between 1 and 10
num1 = random.randint(1, 10)
num2 = random.randint(1, 10)

# Ask user to calculate the product
print("What is the product of", num1, "and", num2, "?")
user_answer = int(input("Enter your answer: "))

# Calculate correct answer
correct_answer = num1 * num2

# Check answer
if user_answer == correct_answer:
    print("Correct! Your answer is right.")
else:
    print("Incorrect. The correct answer is:", correct_answer)
````

### Explanation

- The `random` module is used to generate random numbers.
- The user enters the product.
- The program compares the user's answer with the correct product.
- Based on the result, it displays whether the answer is correct or incorrect.


## Exercise 3: Squares of Even Numbers

**Problem:**  
Create a Python script that prints the squares of all even or odd numbers within the range of 100 to 200.

**Choice:**  
Here, I have chosen **even numbers**.


````python
# Exercise 3: Squares of Even Numbers from 100 to 200

# I have chosen even numbers for this exercise

print("Squares of even numbers from 100 to 200:")

for num in range(100, 201):
    if num % 2 == 0:
        square = num ** 2
        print(num, ":", square)
````

### Explanation

- The loop runs from 100 to 200.
- A number is even if it is divisible by 2.
- For each even number, the square is calculated using `num ** 2`.


## Exercise 4: Word Counter

**Problem:**  
Write a program to count the number of words in a given text.

Example input:

```python
input_text = "This is a sample text. This text will be used to demonstrate the word counter."
```


````python
# Exercise 4: Word Counter

# Given input text
input_text = "This is a sample text. This text will be used to demonstrate the word counter."

# Split the text into words
words = input_text.split()

# Create an empty dictionary to store word counts
word_count = {}

# Count each word
for word in words:
    if word in word_count:
        word_count[word] += 1
    else:
        word_count[word] = 1

# Display the word count
print("Word Count:")
for word, count in word_count.items():
    print(f"'{word}': {count}")
````

### Explanation

- The `split()` function breaks the sentence into words.
- A dictionary is used to store each word and its count.
- If a word appears again, its count is increased by 1.


## Exercise 5: Check for Palindrome

**Problem:**  
Write a Python function called `is_palindrome` that takes a string as input and returns `True` if the string is a palindrome, and `False` otherwise.

A palindrome reads the same forward and backward, ignoring spaces, punctuation, and capitalization.

Example:

Input: `"racecar"`  
Expected Output: `True`


````python
# Exercise 5: Check for Palindrome

import string

def is_palindrome(text):
    """
    This function checks whether the given text is a palindrome or not.
    It ignores spaces, punctuation, and capitalization.
    """

    # Convert text to lowercase
    text = text.lower()

    # Remove spaces and punctuation
    cleaned_text = ""
    for char in text:
        if char.isalnum():
            cleaned_text += char

    # Check if cleaned text is equal to its reverse
    return cleaned_text == cleaned_text[::-1]


# Test the function
input_text = input("Enter a word or phrase: ")

if is_palindrome(input_text):
    print("True - It is a palindrome")
else:
    print("False - It is not a palindrome")
````

### Explanation

- The input string is converted to lowercase.
- Spaces and punctuation are removed.
- The cleaned string is compared with its reverse.
- If both are the same, the string is a palindrome.
