# Hypothesis Testing Assignment

## Topic
Testing whether weekly operating costs are higher than the theoretical cost model.

## Objective
To investigate the restaurant owners' claim that weekly operating costs are higher than the model suggests.


```python
# Install required library if needed
# !pip install scipy numpy
```

## Step 1: Import Required Libraries

```python
import numpy as np
from scipy.stats import norm
```

## Step 2: Given Information

The weekly operating cost model is:

**W = 1000 + 5X**

Where:

- W = weekly operating cost
- X = number of units produced in a week

Given:

- Sample size, n = 25
- Sample mean weekly cost, x̄ = 3050
- Mean units produced, μX = 600
- Standard deviation of units produced, σX = 25
- Significance level, α = 0.05

Since W = 1000 + 5X:

Theoretical mean weekly cost:

**μ = 1000 + 5(600) = 4000**

Standard deviation of weekly cost:

**σ = 5 × 25 = 125**


```python
# Given values
sample_mean = 3050
n = 25
mu_X = 600
sigma_X = 25
alpha = 0.05

# Theoretical mean weekly cost according to model
population_mean = 1000 + 5 * mu_X

# Standard deviation of weekly cost
population_std = 5 * sigma_X

print('Sample Mean:', sample_mean)
print('Theoretical Population Mean:', population_mean)
print('Population Standard Deviation:', population_std)
print('Sample Size:', n)
print('Alpha:', alpha)
```

## Step 3: State the Hypotheses

The restaurant owners claim that the weekly operating costs are higher than the model suggests.

### Null Hypothesis (H₀)
The mean weekly operating cost is less than or equal to the theoretical mean cost.

**H₀: μ ≤ 4000**

### Alternative Hypothesis (H₁)
The mean weekly operating cost is higher than the theoretical mean cost.

**H₁: μ > 4000**

This is a **right-tailed test** because we are checking whether the cost is higher.


## Step 4: Calculate Test Statistic

Formula:

**Z = (x̄ - μ) / (σ / √n)**

Where:

- x̄ = sample mean
- μ = theoretical population mean
- σ = population standard deviation
- n = sample size


```python
# Calculating Z test statistic
z_statistic = (sample_mean - population_mean) / (population_std / np.sqrt(n))

print('Z Test Statistic:', z_statistic)
```

## Step 5: Determine Critical Value

Since this is a right-tailed test at α = 0.05, the critical value is taken from the standard normal distribution.


```python
# Critical value for right-tailed test
critical_value = norm.ppf(1 - alpha)

print('Critical Value:', critical_value)
```

## Step 6: Calculate P-value

```python
# P-value for right-tailed test
p_value = 1 - norm.cdf(z_statistic)

print('P-value:', p_value)
```

## Step 7: Decision Rule

- If Z statistic > Critical Value, reject H₀.
- If Z statistic ≤ Critical Value, fail to reject H₀.

Also:

- If p-value < 0.05, reject H₀.
- If p-value ≥ 0.05, fail to reject H₀.


```python
if z_statistic > critical_value:
    print('Decision based on critical value: Reject the Null Hypothesis')
else:
    print('Decision based on critical value: Fail to Reject the Null Hypothesis')

if p_value < alpha:
    print('Decision based on p-value: Reject the Null Hypothesis')
else:
    print('Decision based on p-value: Fail to Reject the Null Hypothesis')
```

## Step 8: Conclusion

The calculated Z statistic is compared with the critical value at 5% significance level.

Here, the sample mean cost is 3050 while the theoretical mean cost from the model is 4000. The calculated Z statistic is negative, which means the sample mean is lower than the theoretical mean, not higher.

### Final Conclusion
We fail to reject the null hypothesis.

There is not enough statistical evidence to support the restaurant owners' claim that weekly operating costs are higher than the model suggests.


## Final Assignment Answer Summary

- Significance level: 0.05
- Sample mean: 3050
- Theoretical mean: 4000
- Population standard deviation: 125
- Sample size: 25
- Decision: Fail to reject the null hypothesis
- Conclusion: There is no strong evidence that weekly operating costs are higher than the model suggests.
