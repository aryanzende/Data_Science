# Chi-Square Test Assignment

## Topic
Association between Device Type and Customer Satisfaction

## Objective
To check whether there is a significant association between the type of smart home device purchased and customer satisfaction level.


```python
# Install required library if needed
# !pip install scipy pandas numpy
```

## Step 1: Import Required Libraries

```python
import numpy as np
import pandas as pd
from scipy.stats import chi2, chi2_contingency
```

## Step 2: Create the Contingency Table

The given data contains two device types:

- Smart Thermostat
- Smart Light

And five customer satisfaction levels.


```python
# Creating the observed frequency table
data = {
    'Smart Thermostat': [50, 80, 60, 30, 20],
    'Smart Light': [70, 100, 90, 50, 50]
}

satisfaction_levels = [
    'Very Satisfied',
    'Satisfied',
    'Neutral',
    'Unsatisfied',
    'Very Unsatisfied'
]

observed_df = pd.DataFrame(data, index=satisfaction_levels)
observed_df
```

## Step 3: State the Hypotheses

### Null Hypothesis (H₀)
There is no significant association between device type and customer satisfaction level.

In simple words, customer satisfaction is independent of the type of device purchased.

### Alternative Hypothesis (H₁)
There is a significant association between device type and customer satisfaction level.

In simple words, customer satisfaction depends on the type of device purchased.


## Step 4: Perform Chi-Square Test

```python
# Performing Chi-Square test of independence
chi_square_statistic, p_value, dof, expected = chi2_contingency(observed_df)

print('Chi-Square Statistic:', chi_square_statistic)
print('P-value:', p_value)
print('Degrees of Freedom:', dof)

expected_df = pd.DataFrame(expected, index=satisfaction_levels, columns=observed_df.columns)
print('Expected Frequencies:')
display(expected_df)
```

## Step 5: Calculate Critical Value

Formula for degrees of freedom:


**df = (number of rows - 1) × (number of columns - 1)**

Here:

- Rows = 5 satisfaction categories
- Columns = 2 device types

So:

**df = (5 - 1) × (2 - 1) = 4**

Significance level:

**α = 0.05**


```python
alpha = 0.05
critical_value = chi2.ppf(1 - alpha, dof)

print('Significance Level:', alpha)
print('Degrees of Freedom:', dof)
print('Critical Value:', critical_value)
```

## Step 6: Decision Rule

- If Chi-Square Statistic > Critical Value, reject H₀.
- If Chi-Square Statistic ≤ Critical Value, fail to reject H₀.

Also:

- If p-value < 0.05, reject H₀.
- If p-value ≥ 0.05, fail to reject H₀.


```python
if chi_square_statistic > critical_value:
    print('Decision based on critical value: Reject the Null Hypothesis')
else:
    print('Decision based on critical value: Fail to Reject the Null Hypothesis')

if p_value < alpha:
    print('Decision based on p-value: Reject the Null Hypothesis')
else:
    print('Decision based on p-value: Fail to Reject the Null Hypothesis')
```

## Step 7: Conclusion

Based on the Chi-Square test, we compare the calculated Chi-Square statistic with the critical value at 5% significance level.

If the Chi-Square statistic is less than the critical value and the p-value is greater than 0.05, we fail to reject the null hypothesis.

### Final Conclusion
There is not enough statistical evidence to conclude that device type and customer satisfaction level are significantly associated.

Therefore, customer satisfaction appears to be independent of whether the customer purchased a Smart Thermostat or Smart Light.


## Final Assignment Answer Summary

- Test used: Chi-Square Test of Independence
- Significance level: 0.05
- Degrees of freedom: 4
- Null hypothesis: Device type and satisfaction are independent
- Alternative hypothesis: Device type and satisfaction are associated
- Decision: Fail to reject the null hypothesis
- Conclusion: No significant association exists between device type and customer satisfaction level.
